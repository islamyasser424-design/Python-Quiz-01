```python
total_spent = 1200.0
promo_code = " save15 "
is_vip = False

base_discount_rate = 0.0
if total_spent >= 1000: base_discount_rate = 0.10

promo_bonus = 0.0
if promo_code.strip().lower() == "save15": promo_bonus = 0.15

total_discount_rate = base_discount_rate + promo_bonus
discount_amount = total_spent * total_discount_rate

shipping_fee = 0.0
if not is_vip: shipping_fee = 50.0

final_total = total_spent - discount_amount + shipping_fee

print("=== CUSTOMER CHECKOUT SUMMARY ===")
print(f"Original Amount : EGP {total_spent:.2f}")
print(f"Discount Rate   : {total_discount_rate * 100}%")
print(f"Discount Amount : EGP {discount_amount:.2f}")
print(f"Shipping Fee    : EGP {shipping_fee:.2f}")
print(f"Final Total     : EGP {final_total:.2f}")
```

```python
raw_user_record = " 8801 | mohamed ali | 24 "

fields = raw_user_record.split("|")
user_id = fields[0].strip()
full_name = fields[1].strip().title()
age_str = fields[2].strip()

application_status = ""
age_value = 0

try: age_value = int(age_str)
except ValueError: application_status = "ERROR: Invalid Age Format"

if age_value >= 21 and application_status == "": application_status = "APPROVED"
elif age_value < 21 and application_status == "": application_status = "REJECTED: Underage"

print("=== Application VERIFICATION ===")
print(f"User ID      : {user_id}")
print(f"Full Name    : {full_name}")
print(f"Age          : {age_value}")
print(f"Application  : {application_status}")
```

```python
base_salary = 15000.0
sales_target = 100000.0
sales_achieved = 125000.0
csat_score = 4.8

achievement_rate = sales_achieved / sales_target
performance_rating = ""
bonus_percentage = 0.0

if achievement_rate >= 1.20 and csat_score >= 4.5: performance_rating = "Exceeds Expectations"; bonus_percentage = 0.20
elif achievement_rate >= 1.00: performance_rating = "Meets Expectations"; bonus_percentage = 0.10
else: performance_rating = "Needs Improvement"; bonus_percentage = 0.0

calculated_bonus = base_salary * bonus_percentage
if csat_score < 3.0: calculated_bonus = calculated_bonus * 0.5
total_payout = base_salary + calculated_bonus

print("=== EMPLOYEE PAYROLL EVALUATION ===")
print(f"Achievement Rate : {achievement_rate * 100}%")
print(f"Performance Tier : {performance_rating}")
print(f"Calculated Bonus : EGP {calculated_bonus:.2f}")
print(f"Total Payout     : EGP {total_payout:.2f}")
```
