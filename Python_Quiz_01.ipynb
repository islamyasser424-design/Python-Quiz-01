```python
total_spent = 1200.0
promo_code = " save15 "
is_vip = False

base_discount_rate = 0.0
if total_spent >= 1000: base_discount_rate = 0.10

promo_bonus = 0.0
cleaned_promo = promo_code.strip().lower()
if cleaned_promo == "save15": promo_bonus = 0.15

total_discount_rate = base_discount_rate + promo_bonus
discount_amount = total_spent * total_discount_rate

shipping_fee = 0.0
if not is_vip: shipping_fee = 50.0

final_total = total_spent - discount_amount + shipping_fee

print("=== CUSTOMER CHECKOUT SUMMARY ===")
print("Original Amount : EGP " + str(total_spent))
print("Discount Rate   : " + str(total_discount_rate * 100) + "%")
print("Discount Amount : EGP " + str(discount_amount))
print("Shipping Fee    : EGP " + str(shipping_fee))
print("Final Total     : EGP " + str(final_total))
```

```python
raw_user_record = " 8801 | mohamed ali | 24 "

fields = raw_user_record.split("|")
user_id = fields[0].strip()
full_name = fields[1].strip().title()
age_str = fields[2].strip()

age_value = 0
application_status = ""

is_valid_age = age_str.isdigit()
if is_valid_age: age_value = int(age_str)
if is_valid_age and age_value >= 21: application_status = "APPROVED"
if is_valid_age and age_value < 21: application_status = "REJECTED: Underage"
if not is_valid_age: application_status = "ERROR: Invalid Age Format"

print("=== APPLICANT VERIFICATION ===")
print("User ID      : " + str(user_id))
print("Full Name    : " + full_name)
print("Age          : " + str(age_value))
print("Application  : " + application_status)
```

```python
base_salary = 15000.0
sales_target = 100000.0
sales_achieved = 125000.0
csat_score = 4.8

achievement_rate = sales_achieved / sales_target

performance_rating = "Needs Improvement"
bonus_percentage = 0.0

if achievement_rate >= 1.00: performance_rating = "Meets Expectations"
if achievement_rate >= 1.00: bonus_percentage = 0.10
if achievement_rate >= 1.20 and csat_score >= 4.5: performance_rating = "Exceeds Expectations"
if achievement_rate >= 1.20 and csat_score >= 4.5: bonus_percentage = 0.20

calculated_bonus = base_salary * bonus_percentage
if csat_score < 3.0: calculated_bonus = calculated_bonus * 0.5

total_payout = base_salary + calculated_bonus

print("=== EMPLOYEE PAYROLL EVALUATION ===")
print("Achievement Rate : " + str(achievement_rate * 100) + "%")
print("Performance Tier : " + performance_rating)
print("Calculated Bonus : EGP " + str(calculated_bonus))
print("Total Payout     : EGP " + str(total_payout))
```
